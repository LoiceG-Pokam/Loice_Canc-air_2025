```python
import pandas as pd
import seaborn as sns 
import plotly.express as px
import matplotlib.pyplot as plt
import numpy as np
```

## Analyse saisonnalité diagnostic toute pathologie

```python
df_clinique = pd.read_csv("Resultats/Excels/patients_FR_geocoded_adultes_clinique.csv",sep=";", dtype={"codepost":str}, parse_dates=["date_diag"])
df_clinique['year'] = df_clinique["date_diag"].dt.year
df_clinique['month'] = df_clinique["date_diag"].dt.month
```

```python
df_clinique = df_clinique[["pseudo_provisoire","year","month"]]
df_clinique
```

```python

df_clinique_trunc = df_clinique.groupby(['year','month']).count().reset_index()

df_clinique_trunc = df_clinique_trunc.sort_values(by=['year','month'])

df_clinique_trunc["month_name"] = df_clinique_trunc['month'].apply(lambda x: pd.to_datetime(x,format='%m').month_name())

df_clinique_trunc = df_clinique_trunc[["year","month_name","pseudo_provisoire"]]

df_clinique_trunc = df_clinique_trunc.rename({'pseudo_provisoire':'count'},axis=1)
# df_clinique_trunc
```

```python
df_clinique_trunc[df_clinique_trunc['year']>=2017]
```

```python
df_clinique_trunc = df_clinique_trunc.iloc[504:]

# Création du graphique
fig = px.line(df_clinique_trunc, x='month_name', y='count', color='year',
              labels={'count':'Nombre d\'hospitalisations', 'month_name':'Mois', 'year':'Année'},
              title='Analyse saisonnière des dates de diagnostic par année')

fig.update_xaxes(categoryorder='array', categoryarray= [pd.to_datetime(x, format='%B').month_name() for x in pd.date_range('2020-01-01', periods=12, freq='M').month_name()])
fig.update_layout(xaxis_title='Mois', yaxis_title='Nombre d\'hospitalisations', legend_title='Année')
fig.show()
```

## Analyse saisonnalité evenements médicaux 

```python
df_ap = pd.read_excel("../geocodeur/from_hegp/data/data_octobre_2023/medical_event/aplasie.xlsx", parse_dates=True)
```

```python
df_ap['year'] = df_ap["Date début hospitalisation"].dt.year
df_ap['month'] = df_ap["Date début hospitalisation"].dt.month
df_ap['periode'] = df_ap["Date fin hospitalisation"] - df_ap["Date début hospitalisation"]

df_ap_yrs = df_ap.groupby(['year','month']).count().reset_index()
df_ap_yrs = df_ap_yrs.iloc[1:]

df_ap_yrs = df_ap_yrs.sort_values(by=['year','month'])

df_ap_yrs["month_name"] = df_ap_yrs['month'].apply(lambda x: pd.to_datetime(x,format='%m').month_name())

df_ap_yrs_truncated = df_ap_yrs[["year","month_name","pseudo_provisoire"]]

df_ap_yrs_truncated = df_ap_yrs_truncated.rename({'pseudo_provisoire':'count'},axis=1)
```

### Analyse de la saisonnalité - par année : 

```python
# Création du graphique
fig = px.line(df_ap_yrs_truncated, x='month_name', y='count', color='year',
              labels={'count':'Nombre d\'hospitalisations', 'month_name':'Mois', 'year':'Année'},
              title='Analyse saisonnière des hospitalisations pour aplasie par année')

fig.update_xaxes(categoryorder='array', categoryarray= [pd.to_datetime(x, format='%B').month_name() for x in pd.date_range('2020-01-01', periods=12, freq='M').month_name()])
fig.update_layout(xaxis_title='Mois', yaxis_title='Nombre d\'hospitalisations', legend_title='Année')
fig.show()
```
