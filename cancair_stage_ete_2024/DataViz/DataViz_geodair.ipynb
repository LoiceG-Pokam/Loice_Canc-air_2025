```python
import pandas as pd
import plotly.express as px
import geopandas as gpd
import plotly.graph_objects as go
import json
import numpy as np 
from dateutil.parser import parse
import datetime
```

```python
df_pm10 = pd.read_csv("../data/geodair/Moy_jour_PM10_2023_01_01-2023_11_09.csv", sep = ";")
```

```python
col = df_pm10.columns
#df_pm10["date"] = df_pm10["Date de début"]
col_to_keep = ["Date de début", 'code site', "type d'influence", 'valeur', 'valeur brute','couverture de données','code qualité', 'validité', 'Latitude', 'Longitude']

df_pm10 = df_pm10[col_to_keep]

df_pm10['date'] = [parse(date).date() for date in df_pm10["Date de début"]]
df_pm10['monthyear'] = pd.to_datetime(df_pm10['date']).dt.to_period('M')

df_pm10.drop("Date de début", axis=1)
```

```python
## Visualisation pour une date donnée : 

df_pm10['date'] = pd.to_datetime(df_pm10['date'])  

mask = (df_pm10['date'] == '2023-11-09')    

df_0911 = df_pm10.loc[mask]
```

```python
gdf = (gpd.GeoDataFrame(df_0911, geometry=gpd.points_from_xy(df_0911.Longitude, df_0911.Latitude)))
```

```python
df_0911
```

```python
fig = px.scatter_mapbox(gdf, lat='Latitude', lon='Longitude', color='valeur',
                            color_continuous_scale="bluered",
                            size_max=15, zoom=7,
                            mapbox_style="carto-positron",
                            labels={'valeur': 'valeur'})
fig.show()
```
